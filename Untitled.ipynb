```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#https://github.com/sdryan/AntsMultipleFoodSources/blob/main/ants_latticecode_SR_twofoodsources_ne.m

dt=0.001
dx=1.0
N=50    #lattice size
M=100  #number of ants
Mf=2    #food source
D=10.0  #diffusion
gamma=0.001   #evaporation coefficient
CFL=D*dt/(dx*dx)    #CFL condition
q=1.0   #ant pheromone production rate
time=2.0
count1=0
Track=0
```

```python
#build lattice
x=np.zeros((N+1, N+1))
y=np.zeros((N+1, N+1))

for i in range(N+1):
    for j in range(N+1):
        x[i,j]=i*dx
        y[i,j]=j*dx
```

```python
#IC on pheromone
c=np.zeros((N+1, N+1, 2))  #pheromone
dens=np.zeros((N+1, N+1))
dens_ret=np.zeros((N+1, N+1))

for i in range(N+1):
    for j in range(N+1):
        c[i,j,0]=0
        c[i,j,1]=0
        dens[i,j]=0
        dens_ret[i,j]=0
```

```python
#place ants at nest y
xn=[25,25]
ant=np.zeros((M, 2), dtype=np.int64)

for i in range(M):
    ant[i]=xn    #place nest at center
```

```python
#food source location
xf=np.array([[5,45],
             [25,5]])

mark=np.zeros(M)    #0 for foragers and 1 for returners
fs=np.zeros((M,2))
```

```python
#time loop
food_found=0

for k in range(int(time/dt)):
    #check if food has been found
    for p in range(M):
        for p1 in range(Mf):
            if ant[p,0]==xf[p1,0] and ant[p,1]==xf[p1,1]:
                food_found=1

    #convert forager to returner
    for i in range(M):
        if mark[i]==0:
            for p1 in range(Mf):
                dist1=np.linalg.norm(ant[i]-xf[p1])
                if dist1<=dx:     #if you are next to the food you now return
                    mark[i]=1
                    fs[i]=xf[p1]
        else:
            dist1=np.linalg.norm(ant[i]-xn)   #if you are back at the nest become a forager
            if dist1<=dx:
                mark[i]=0   #change to forager
                
    if food_found==1:
        #solve PDE for c(x,t) with BCs
        for i in range(1,N):
            for j in range(1,N):
                c[i,j,1]= c[i,j,0] + (D*dt/(dx*dx))*(c[i+1,j,0] + c[i-1,j,0] - 4.0*c[i,j,0] + c[i,j+1,0] + c[i,j-1,0]) - gamma*dt*c[i,j,0]

        #add ant pheromone contribution
        for p in range(M):
            if mark[p]:  #returner i.e. have found food
                c[ant[p,0], ant[p,1], 1]+=dt*q*np.exp(-((ant[p,0]-fs[p,0])**2 + (ant[p,1]-fs[p,1])**2))

        #boundary conditions using Robin condition
        for i in range(N+1):
            c[0,i,1] = c[1,i,0]-dx/D*c[0,i,0]
            c[N,i,1] = c[N-1,i,0]-dx/D*c[N,i,0]
            c[i,0,1] = c[i,1,0]-dx/D*c[i,0,0]
            c[i,N,1] = c[i,N-1,0]-dx/D*c[i,N,0]

        #rest for next time step
        for i in range(N+1):
            for j in range(N+1):
                if c[i,j,1]<0.00000000001:
                    c[i,j,1]=0   #ensure that pheromone conc is nonnegative
                c[i,j,0]=c[i,j,1]

    #ant dynamics
    wa=np.zeros(8)

    for i in range(M):
        #returner dynamics
        if mark[i]:   
            vec_home=xn-ant[i]   #beeline
            distance=np.linalg.norm(vec_home)

            ant[i]+=np.int64(1.0*vec_home/distance if distance>0 else ant[i].copy())

        #forager dynamics
        else:
            #compute prob for 8 neighboring cells
            curr_c=c[ant[i,0], ant[i,1], 1]
            
            #neighbor 1
            local_c=c[ant[i,0]-1, ant[i,1]+1, 1] - curr_c   
            if local_c<0:
                wa[0]=0
            elif local_c>0:
                wa[0]=1+local_c
            else:
                wa[0]=1

            #neighbor 2
            local_c=c[ant[i,0], ant[i,1]+1, 1] - curr_c
            if local_c<0:
                wa[1]=0
            elif local_c>0:
                wa[1]=1+local_c
            else:
                wa[1]=1

            #neighbor 3
            local_c=c[ant[i,0]+1, ant[i,1]+1, 1] - curr_c
            if local_c<0:
                wa[2]=0
            elif local_c>0:
                wa[2]=1+local_c
            else:
                wa[2]=1

            #neighbor 4
            local_c=c[ant[i,0]+1, ant[i,1], 1] - curr_c
            
            if local_c<0:
                wa[3]=0
            elif local_c>0:
                wa[3]=1+local_c
            else:
                wa[3]=1

            #neighbor5
            local_c=c[ant[i,0]+1, ant[i,1]-1, 1] - curr_c
            if local_c<0:
                wa[4]=0
            elif local_c>0:
                wa[4]=1+local_c
            else:
                wa[4]=1

            #neighbor6
            local_c=c[ant[i,0], ant[i,1]-1, 1] - curr_c
            if local_c<0:
                wa[5]=0
            elif local_c>0:
                wa[5]=1+local_c
            else:
                wa[5]=1

            #neighbor7
            local_c=c[ant[i,0]-1, ant[i,1]-1, 1] - curr_c
            if local_c<0:
                wa[6]=0
            elif local_c>0:
                wa[6]=1+local_c
            else:
                wa[6]=1

            #neighbor8
            local_c=c[ant[i,0]-1, ant[i,1], 1] - curr_c
            if local_c<0:
                wa[7]=0
            elif local_c>0:
                wa[7]=1+local_c
            else:
                wa[7]=1

            wa_sum=np.sum(wa)

            #choose random numebr [0,1]
            r1=np.random.uniform(low=0.0, high=1.0, size=None)

            #move neighbors
            if r1<wa[0]/wa_sum:
                #move to neighbor 1
                ant[i]+=np.array([-1,1])
            elif r1<np.sum(wa[:2])/wa_sum:
                ant[i]+=np.array([0,1])
            elif r1<np.sum(wa[:3])/wa_sum:
                ant[i]+=np.array([1,1])
            elif r1<np.sum(wa[:4])/wa_sum:
                ant[i]+=np.array([1,0])
            elif r1<np.sum(wa[:5])/wa_sum:
                ant[i]+=np.array([1,-1])
            elif r1<np.sum(wa[:6])/wa_sum:
                ant[i]+=np.array([0,-1])
            elif r1<np.sum(wa[:7])/wa_sum:
                ant[i]+=np.array([-1,-1])
            else:
                ant[i]+=np.array([-1,0])

            #reflecting BCs
            if ant[i,0]==0:
                ant[i,0]=1
            if ant[i,0]==N:
                ant[i,0]=N-1
            if ant[i,1]==0:
                ant[i,1]=1
            if ant[i,1]==N:
                ant[i,1]=N-1

    #plot every so many frames
    count1+=1
    if count1>9:
        count1=0

        #density plot
        dens=np.zeros((N+1, N+1))
        dens_ret=np.zeros((N+1, N+1))

        for p in range(M):
            if not mark[p]:
                dens[ant[p,0], ant[p,1]]+=1
            else:
                dens_ret[ant[p,0], ant[p,1]]+=1

        #scaling for images
        c1=c[:,:,1]

    
        # Create three panels.
        fig, axes = plt.subplots(1, 3, figsize=(18, 5), constrained_layout=True)
        
        # MATLAB labels run from 1 through N+1.
        # Show only labels divisible by 5.
        tick_positions = np.arange(4, N + 1, 5)
        tick_labels = (tick_positions + 1).astype(str)
        
        # "Blues" is used as an approximation to MATLAB's "sky" colormap.
        panels = [
            (dens, "Forager Distribution", "hot"),
            (dens_ret, "Returner Distribution", "Blues"),
            (c1, "Pheromone Distribution", "Blues"),
        ]
        
        for ax, (data, title, cmap) in zip(axes, panels):
            image = ax.imshow(
                data.T,
                origin="lower",
                cmap=cmap,
                interpolation="nearest",
                aspect="equal",
            )
        
            ax.set_title(title)
            ax.set_xticks(tick_positions)
            ax.set_xticklabels(tick_labels)
            ax.set_yticks(tick_positions)
            ax.set_yticklabels(tick_labels)
            ax.grid(False)
        
            fig.colorbar(image, ax=ax)
        
        # Refresh the figure, analogous to MATLAB's drawnow.
        plt.draw()
        plt.pause(0.001)
        
        # Increment the output counter.
        Track += 1

        '''
        # Save the original arrays, not their transposes.
        np.savetxt(
            f"forager_dens_{Track:05d}.dat",
            dens,
            delimiter="\t",
            fmt="%.5g",
        )
        np.savetxt(
            f"returner_dens_{Track:05d}.dat",
            dens_ret,
            delimiter="\t",
            fmt="%.5g",
        )
        np.savetxt(
            f"pheromone_dens_{Track:05d}.dat",
            c1,
            delimiter="\t",
            fmt="%.5g",
        )
        '''
        # Optional: save the figure.
        # fig.savefig(f"chrom_mov_{Track:05d}.jpg", dpi=150)
        
        plt.show()
```

```python
import math 
N=50
c=np.zeros((N+1, N+1, 2))  #pheromone
dx=1.0

i,j=20,30

rng=np.random.default_rng()
wind_vel=rng.random(1)*20+5
wind_vec=rng.random(2)*10-5 
unit_wind_vec=wind_vec/np.linalg.norm(wind_vec)
scaled_wind_vec=unit_wind_vec*wind_vel

print(wind_vel)
print(unit_wind_vec)
print(scaled_wind_vec)

#get area of affect
di=math.ceil(scaled_wind_vec[0])
dj=math.ceil(scaled_wind_vec[1])

i_start = max(0, min(i, i + di))
j_start = max(0, min(j, j + dj))

i_stop = min(c.shape[0], max(i, i + di) + int(di < 0))
j_stop = min(c.shape[1], max(j, j + dj) + int(dj < 0))

c[i_start:i_stop, j_start:j_stop, 0] = 1

for xi in range(i_start, i_stop):
    for xj in range(j_start, j_stop):
        c[xi,xj,0]-=np.exp(-np.linalg.norm(np.array([xi-i,xj-j]))*dx/wind_vel[0]) * c[xi,xj,0]
        c[xi,xj,0]=max(0.001, c[xi,xj,0])

print(i_start, i_stop, j_start, j_stop)
print(c[i_start:i_stop, j_start:j_stop, 0])
```
